```python
%sql
-- Validation for instacart.instacart_raw.aisles

SELECT
    'aisles' AS table_name,
    COUNT(*) AS total_rows,

    -- Duplicate ID count
    COUNT(*) 
        - COUNT(DISTINCT aisle_id)
        - SUM(CASE
                WHEN aisle_id IS NULL THEN 1
                ELSE 0
              END) AS duplicate_aisle_id_count,

    -- NULL counts
    SUM(CASE
            WHEN aisle_id IS NULL THEN 1
            ELSE 0
        END) AS aisle_id_null_count,

    SUM(CASE
            WHEN aisle IS NULL THEN 1
            ELSE 0
        END) AS aisle_null_count,

    -- Missing value count
    SUM(CASE
            WHEN LOWER(TRIM(aisle)) = 'missing' THEN 1
            ELSE 0
        END) AS aisle_missing_count,

    -- NULL percentages
    ROUND(
        100.0 * SUM(CASE
            WHEN aisle_id IS NULL THEN 1
            ELSE 0
        END) / COUNT(*),
        2
    ) AS aisle_id_null_pct,

    ROUND(
        100.0 * SUM(CASE
            WHEN aisle IS NULL THEN 1
            ELSE 0
        END) / COUNT(*),
        2
    ) AS aisle_null_pct,

    -- Missing percentage
    ROUND(
        100.0 * SUM(CASE
            WHEN LOWER(TRIM(aisle)) = 'missing' THEN 1
            ELSE 0
        END) / COUNT(*),
        2
    ) AS aisle_missing_pct

FROM instacart.instacart_raw.aisles;
```

```python
%sql
-- Validation for instacart.instacart_raw.order_products_prior

SELECT
    'order_products_prior' AS table_name,
    COUNT(*) AS total_rows,

    -- Duplicate business key: order_id + product_id
    (
        SELECT COALESCE(SUM(duplicate_count - 1), 0)
        FROM (
            SELECT
                order_id,
                product_id,
                COUNT(*) AS duplicate_count
            FROM instacart.instacart_raw.order_products_prior
            GROUP BY order_id, product_id
            HAVING COUNT(*) > 1
        )
    ) AS duplicate_order_product_count,

    -- NULL counts
    SUM(CASE
            WHEN order_id IS NULL THEN 1
            ELSE 0
        END) AS order_id_null_count,

    SUM(CASE
            WHEN product_id IS NULL THEN 1
            ELSE 0
        END) AS product_id_null_count,

    SUM(CASE
            WHEN add_to_cart_order IS NULL THEN 1
            ELSE 0
        END) AS add_to_cart_order_null_count,

    SUM(CASE
            WHEN reordered IS NULL THEN 1
            ELSE 0
        END) AS reordered_null_count,

    -- NULL percentages
    ROUND(
        100.0 * SUM(CASE
            WHEN order_id IS NULL THEN 1
            ELSE 0
        END) / COUNT(*),
        2
    ) AS order_id_null_pct,

    ROUND(
        100.0 * SUM(CASE
            WHEN product_id IS NULL THEN 1
            ELSE 0
        END) / COUNT(*),
        2
    ) AS product_id_null_pct,

    ROUND(
        100.0 * SUM(CASE
            WHEN add_to_cart_order IS NULL THEN 1
            ELSE 0
        END) / COUNT(*),
        2
    ) AS add_to_cart_order_null_pct,

    ROUND(
        100.0 * SUM(CASE
            WHEN reordered IS NULL THEN 1
            ELSE 0
        END) / COUNT(*),
        2
    ) AS reordered_null_pct

FROM instacart.instacart_raw.order_products_prior;
```

```python
%sql
WITH raw_summary AS (
    SELECT 
        'aisles' AS table_name, 
        COUNT(*) AS total_rows, 
        COUNT(*) - COUNT(DISTINCT struct(*)) AS duplicate_rows
    FROM instacart.instacart_raw.aisles

    UNION ALL

    SELECT 
        'departments', 
        COUNT(*), 
        COUNT(*) - COUNT(DISTINCT struct(*))
    FROM instacart.instacart_raw.departments

    UNION ALL

    SELECT 
        'order_products_prior', 
        COUNT(*), 
        COUNT(*) - COUNT(DISTINCT order_id, product_id)
    FROM instacart.instacart_raw.order_products_prior

    UNION ALL

    SELECT 
        'order_products_train', 
        COUNT(*), 
        COUNT(*) - COUNT(DISTINCT order_id, product_id)
    FROM instacart.instacart_raw.order_products_train

    UNION ALL

    SELECT 
        'orders', 
        COUNT(*), 
        COUNT(*) - COUNT(DISTINCT order_id)
    FROM instacart.instacart_raw.orders

    UNION ALL

    SELECT 
        'products', 
        COUNT(*), 
        COUNT(*) - COUNT(DISTINCT product_id)
    FROM instacart.instacart_raw.products
)

SELECT 
    table_name,
    total_rows,
    duplicate_rows,
    CASE
        WHEN duplicate_rows = 0 THEN 'PASSED' 
        ELSE 'FAILED (Duplicates Found)' 
    END AS validation_status

FROM raw_summary;
```
