```python
%sql
-- validation for instacart.instacart_clean.aisles 
SELECT
    'aisles' AS table_name,
    COUNT(*) AS total_rows,
    COUNT(*) - COUNT(DISTINCT aisle_id) AS duplicate_aisle_id_count,
    
    -- Exact NULL counts
    SUM(CASE 
            WHEN aisle_id IS NULL THEN 1 
            ELSE 0 
        END) AS aisle_id_null_count,

    SUM(CASE 
            WHEN aisle IS NULL THEN 1 
            ELSE 0 
        END) AS aisle_null_count,
    
    -- NULL Percentages
    ROUND(100.0 * SUM
        (CASE 
            WHEN aisle_id IS NULL THEN 1 
            ELSE 0 
        END) / COUNT(*), 2) AS aisle_id_null_pct,

    ROUND(100.0 * SUM
        (CASE 
            WHEN aisle IS NULL THEN 1 
            ELSE 0 
        END) / COUNT(*), 2) AS aisle_null_pct
FROM instacart.instacart_clean.aisles;
```

```python
%sql
WITH clean_summary AS (
    SELECT 
        'aisles' AS table_name, 
        COUNT(*) AS total_rows, 
        COUNT(*) - COUNT(DISTINCT struct(*)) AS duplicate_rows      -- counts the number of duplicate rows 
    FROM instacart.instacart_clean.aisles

    UNION ALL

    SELECT 
        'departments', 
        COUNT(*), 
        COUNT(*) - COUNT(DISTINCT struct(*))        
    FROM instacart.instacart_clean.departments

    UNION ALL

    SELECT 
        'order_products_prior', 
        COUNT(*), 
        COUNT(*) - COUNT(DISTINCT order_id, product_id) 
    FROM instacart.instacart_clean.order_products_prior

    UNION ALL

    SELECT 
        'order_products_train', 
        COUNT(*), 
        COUNT(*) - COUNT(DISTINCT order_id, product_id) 
    FROM instacart.instacart_clean.order_products_train

    UNION ALL

    SELECT 
        'orders', 
        COUNT(*), 
        COUNT(*) - COUNT(DISTINCT order_id) 
    FROM instacart.instacart_clean.orders

    UNION ALL
    SELECT 
        'products', 
        COUNT(*), 
        COUNT(*) - COUNT(DISTINCT product_id) 
    FROM instacart.instacart_clean.products
)

SELECT 
    table_name,
    total_rows,
    duplicate_rows,
    CASE
        WHEN duplicate_rows = 0 THEN 'PASSED' 
        ELSE 'FAILED (Duplicates Found)' 
    END AS validation_status

FROM clean_summary;
```
